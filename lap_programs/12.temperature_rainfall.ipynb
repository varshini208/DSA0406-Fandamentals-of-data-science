1.Python program to create a line plot of monthly temperature data

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

temperature = [25, 27, 30, 34, 36, 35, 32, 31, 30, 29, 27, 24]

plt.plot(months, temperature, marker='o')

plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.title("Monthly Temperature Data")
plt.show()
```

2.Python program to create a scatter plot of monthly rainfall data

```python
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

rainfall = [20, 15, 30, 45, 60, 120, 180, 200, 150, 90, 40, 25]

plt.scatter(months, rainfall)

plt.xlabel("Month")
plt.ylabel("Rainfall (mm)")
plt.title("Monthly Rainfall Data")
plt.show()
```
